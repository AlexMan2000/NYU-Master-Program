```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
N = 1000

def generate_std_gaussian(N):
    return np.random.normal(loc = 0, scale = 1, size = (N,))

def plot_std_gaussian(samples):
    sns.distplot(samples, )

samples = generate_std_gaussian(N)
plot_std_gaussian(samples)
```

```python
def transform(u, w, b, samples):
    return samples + u * np.tanh(samples * w + b)
    
u = 2
w = 3
b = 1
sns.distplot(transform(u, w, b, samples))
```
