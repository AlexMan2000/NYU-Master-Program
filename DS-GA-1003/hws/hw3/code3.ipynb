```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation parameters
T = 3.0  # total time
dt = 0.01  # time step
N = int(T / dt)  # number of time steps
M = 1000  # number of samples to simulate

# Initial condition z0 is a mixture of two normal distributions
z0 = 0.5 * np.random.normal(-10, 1, M) + 0.5 * np.random.normal(10, 1, M)

# Preallocate and initialize the array for the process, simulate them in parallel
zt = np.zeros((M, N+1))
zt[:, 0] = z0

# Simulate the process
for i in range(N):
    zt[:, i+1] = zt[:, i] - 0.5 * zt[:, i] * dt + np.sqrt(dt) * np.random.normal(0, 1, M)

# Plotting just a few paths to illustrate the behavior
plt.figure(figsize=(10, 6))
for i in range(min(10, M)):  # Plot 10 sample paths
    plt.plot(np.linspace(0, T, N+1), zt[i, :], alpha=0.7)



plt.title('Sample Paths of Ornstein-Uhlenbeck Process')
plt.xlabel('Time t')
plt.ylabel('Process z_t')
plt.show()
```
